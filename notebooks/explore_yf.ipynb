```python
# Fetching multiple tickers

# List of tickers you want to test (change these if you want)
tickers = ["AAPL", "MSFT", "GOOGL", "AMZN", "TSLA"]

for ticker in tickers:
    try:
        stock = yf.Ticker(ticker)
        info = stock.info

        name = info.get("longName")
        price = info.get("currentPrice")
        industry = info.get("industry")
        sector = info.get("sector")
        dividend_yield = info.get("dividendYield")
        trailing_pe = info.get("trailingPE")
        forward_pe = info.get("forwardPE")
        average_volume = info.get("averageVolume")
        market_cap = info.get("marketCap")
        enterprise_value = info.get("enterpriseValue")
        price_to_book = info.get("priceToBook")

        # Print some stuff to know how it works
        print(f"{ticker}")
        print("Company Name:", info.get("longName"))
        print("Current Price:", info.get("currentPrice"))
        print("Industry:", info.get("industry"))
        print("Sector:", info.get("sector"))
        print("Dividend Yield:", info.get("dividendYield"))
        print("Trailing PE:", info.get("trailingPE"))
        print("Forward PE:", info.get("forwardPE"))
        print("Average Volume:", info.get("averageVolume"))
        print("Market Cap:", info.get("marketCap"))
        print("Enterprise Value:", info.get("enterpriseValue"))
        print("Price to Book:", info.get("priceToBook"))
        print("-" * 50)

    except Exception as e:
        print(f"Error fetching {ticker}: {e}")
        print("-" * 50)
```
